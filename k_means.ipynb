```python
def distance(p1, p2):
    # both points are np.ndarray of dim 1
    # Euclidian distance
    dim = p1.shape[0]
    sum_square = 0
    for i in range(dim):
        sum_square += (p1[i] - p2[i])**2
    return np.sqrt(sum_square)
```

```python
# putting all together
def k_means(data, k, max_iter = 10):
    # data is np.ndarray with dim 0 of n
    # k is number of clusters
    n = data.shape[0]
    
    # random initialize cluster
    cluster = np.random.randint(0, 2, n)

    # compute centroids
    centroids = {}
    for c in range(k):
        centroids[c] = data[cluster==c].mean(axis=0)

    dists = np.zeros([n, k])
    
    for _ in range(max_iter):
        for i in range(n):
            for c in range(k):
                dists[i,c] = distance(data[i,:], centroids[c])
        
        new_cluster = dists.argmin(axis=1)
        
        if np.all(new_cluster == cluster):
            break
        cluster = new_cluster
    return cluster, centroids
    
def distance(p1, p2):
    # both points are np.ndarray of dim 1
    # Euclidian distance
    dim = p1.shape[0]
    sum_square = 0
    for i in range(dim):
        sum_square += (p1[i] - p2[i])**2
    return np.sqrt(sum_square)
```

```python
cluster, centroids = k_means(data, 2)
```

```python
def k_means(data, k=2, max_iter=100):
    n, d = data.shape
    cluster = np.random.randint(0, k, (n,))
    centroids = {}
    for _ in range(max_iter):
        centroids = {}
        for c in range(d):
            centroids[c] = data[cluster == c, :].mean(axis=0)
    
        dists = np.zeros((n, k))
        for i in range(n):
            for c in range(d):
                dists[i,c] = distance(data[i], centroids[c])
    
        cluster = np.argmin(dists, axis=1)
    return cluster, centroids
```
