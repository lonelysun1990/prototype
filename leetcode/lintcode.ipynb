```python
from typing import (
    List,
)

class Solution:
    """
    @param grid: a boolean 2D matrix
    @return: an integer
    """
    def num_islands(self, grid: List[List[bool]]) -> int:
        from collections import deque
        # edge case
        if grid is None:
            return 0
        
        n_row = len(grid)
        if n_row == 0:
            return 0
        
        n_col = len(grid[0])
        if n_col == 0:
            return 0

        # find num of islands
        visited = [[0]*n_col] * n_row
        island_count = 0
        print("n_row: ", n_row)

        queue = deque([])
        for i_row in range(n_row):
            for i_col in range(n_col):
                if visited[i_row][i_col] == 0 and grid[i_row][i_col] == 1:
                    island_count += 1
                    queue.append((i_row, i_col))
                    visited[i_row][i_col] = 1
                    print("new island location: ", i_row, i_col)
                    
                while len(queue):
                    tile_row, tile_col = queue.popleft()
                    neighbors = self.find_neighbors(grid, visited, tile_row, tile_col)
                    print("neighbors: ", neighbors)
                    for neighbor in neighbors:
                        nei_row, nei_col = neighbor
                        if visited[nei_row][nei_col] == 0:
                            queue.append(neighbor)
                            visited[nei_row][nei_col] = 1
        return island_count
    
    def find_neighbors(
        self, 
        grid: List[List[bool]], 
        visisted: List[List[bool]], 
        idx_row: int, 
        idx_col: int):
        # explore 4 directions
        n_row = len(grid)
        n_col = len(grid[0])

        neighbors = []
        if idx_row < n_row -1 and grid[idx_row+1][idx_col] == 1:
            neighbors.append((idx_row+1, idx_col))
        if idx_row > 0 and grid[idx_row-1][idx_col] == 1:
            neighbors.append((idx_row-1, idx_col))
        if idx_col < n_col -1 and grid[idx_row][idx_col+1] == 1:
            neighbors.append((idx_row, idx_col+1))
        if idx_col > 0 and grid[idx_row][idx_col-1] == 1:
            neighbors.append((idx_row, idx_col-1))
        
        return neighbors
```

```python
from typing import (
    List,
)

class Solution:
    """
    @param nums: A list of integers.
    @return: A list of permutations.
             we will sort your return value in output
    """
    def permute(self, nums: List[int]) -> List[List[int]]:
        if nums is None or len(nums) == 0:
            return [[]]

        visited = [0] * len(nums)
        results = []
        self.dfs(nums, visited, [], results)

    def dfs(self, nums, visited, path, results):
        if sum(visited) == len(nums):
            results.append(path.copy())
            return
        
        for idx in range(len(nums)):
            path.append(nums[idx])
            visited[idx] = 1
            self.dfs(nums, visited, path, results)
            path.pop()
            visited[idx] = 0
```

```python
from typing import (
    List,
)

class Solution:
    """
    @param matrix: the given matrix
    @return: the largest possible sum
    """
    def max_submatrix(self, matrix: List[List[int]]) -> int:
        if matrix is None or len(matrix) == 0 or len(matrix[0]) == 0:
            return 0
        
        # minimum possible sum for matrix[0, i][0, j] (submatrix always starts from [0,0]) 
        min_sum = [[0]*len(matrix[0]) for _ in range(len(matrix))]
        sum_matrix = [[matrix[0][0]]*len(matrix[0]) for _ in range(len(matrix))]

        max_possible_sum = matrix[0][0]

        for i in range(len(matrix)):
            for j in range(len(matrix[0])):
                if i > 0:
                    sum_matrix[i][j] = sum_matrix[i-1][j] + sum([matrix[i][idx_j] for idx_j in range(j+1)])
                    min_sum[i][j] = min(sum_matrix[i][j], min_sum[i][j])
                if j > 0:
                    sum_matrix[i][j] = sum_matrix[i][j-1] + sum([matrix[idx_i][j] for idx_i in range(i+1)])
                    min_sum[i][j] = min(sum_matrix[i][j], min_sum[i][j])
                
                if sum_matrix[i][j] - min_sum[i][j] > max_possible_sum:
                    max_possible_sum = sum_matrix[i][j] - min_sum[i][j]
        print("min_sum: ", min_sum)
        print("sum_matrix: ", sum_matrix)
        
        return max_possible_sum, min_sum, sum_matrix, matrix
```

```python
# 799
from typing import List
def back_pack_vii(value: List[int], amount: List[int], n: int) -> int:
    if len(value) != len(amount):
        return 0
    
    m = len(amount)
    can_do = [1] + [0]*(n) # 0 or 1 representing whether number can be assembled
    for j in range(m):
        count = [0] * (n+1) # count of each type of coin at value n
        for i in range(value[j], n+1):
            if can_do[i] == 0 and can_do[i - value[j]] == 1 and count[i-value[j]] < amount[j]:
                can_do[i] = 1
                count[i] = count[i-value[j]] + 1
    return sum(can_do)
            
    
```

```python
class Solution:
    def word_justification(self, words, max_width):
        results = []
        curr_word_count = 0
        curr_line = []
        curr_space_count = {}
        n = len(words)
        i = 0
        while i < n:
            word = words[i]
            net_length = len(word) + self.compute_curr_width(curr_line, curr_space_count)
            if len(curr_line) == 0 and net_length <= max_width or net_length + 1 <= max_width:
                curr_line.append(word)
                if curr_word_count > 0:
                    curr_space_count[curr_word_count-1] = 1
                i += 1
                curr_word_count += 1
                
            else:
                # add logic to add space
                curr_space_count = self.add_space(curr_line, curr_space_count, max_width, False)
                result_line = self.assemble(curr_line, curr_space_count)
                results.append(result_line)
                curr_word_count = 0
                curr_line = []
                curr_space_count = {}
        
        # add last row
        curr_space_count = self.add_space(curr_line, curr_space_count, max_width, True)
        result_line = self.assemble(curr_line, curr_space_count)
        results.append(result_line)
        return results

    def add_space(self, curr_line, curr_space_count, max_width, if_last_row):
        curr_width = self.compute_curr_width(curr_line, curr_space_count)
        total_space = len(curr_space_count)
        
        # one word in curr_line
        if len(curr_line) == 1 or if_last_row:
            if len(curr_space_count) == 0:
                last_idx = -1
            else:
                last_idx = max([key for key in curr_space_count.keys()])
            curr_space_count[last_idx+1] = max_width - curr_width
            return curr_space_count
        print("curr_line: ", curr_line)
        print("curr_space_count: ", curr_space_count)
        # curr_line has more than 1 word
        next_space = 0
        while curr_width < max_width:
            curr_space_count[next_space] += 1
            next_space = (next_space+1) % total_space 
            curr_width += 1
        return curr_space_count

    def assemble(self, curr_line, curr_space_count):
        result_line = ""
        idx = 0
        while idx < len(curr_line) or idx in curr_space_count:
            if idx < len(curr_line):
                result_line += curr_line[idx]
            if idx in curr_space_count:
                result_line += " " * curr_space_count[idx]
            idx += 1
        return result_line
        

    def compute_curr_width(self, curr_line, curr_space_count):
        total_space = sum([val for key, val in curr_space_count.items()])
        total_text = sum([len(word) for word in curr_line])
        return total_space + total_text
```
