```python
from sklearn.datasets import fetch_california_housing
```

```python
housing = fetch_california_housing(as_frame = True)
```

```python
df = housing.frame
```

```python
X = df.drop('MedHouseVal', axis=1)
y = df['MedHouseVal']
```

```python
X.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_tmp, y_train, y_tmp = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
X_val, X_test, y_val, y_test = train_test_split(X_tmp, y_tmp, test_size=0.5, random_state=42)
```

```python
print(len(X_train))
print(len(X_val))
print(len(X_test))
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
```

```python
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)
```

```python
pd.Series(X_train_scaled.mean(axis=0), index=X.columns).round(5).to_dict()
```

```python
pd.Series(X_val_scaled.mean(axis=0), index=X.columns).round(5).to_dict()
```

```python
pd.Series(X_val_scaled.std(axis=0), index=X.columns).round(4).to_dict()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train_scaled, y_train)
```

```python
y_pred_val = lr.predict(X_val_scaled)
y_pred_test = lr.predict(X_test_scaled)
```

```python
from sklearn.model_selection import cross_validate, KFold
```

```python
cv = KFold(n_splits=5, shuffle=True, random_state=42)
```

```python
pipe_lr = Pipeline([
    ('scaler', StandardScaler()),
    ('reg', LinearRegression())
])
```

```python
cv_result = cross_validate(
    pipe_lr,
    X_train,
    y_train,
    cv=cv,
    scoring=['neg_root_mean_squared_error', 'r2']
)
```

```python
cv_result
```

```python
cv_result['test_neg_root_mean_squared_error']
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
pipe_rf = Pipeline([
    ('scaler', StandardScaler()),
    ('reg', RandomForestRegressor())
])
```

```python
pipe_rf.fit(X_train, y_train)
```

```python
rf_model = pipe_rf.named_steps['reg']
```

```python
importance_native = rf_model.feature_importances_
```

```python
importance_native
```

```python
pd.Series(importance_native, index=X.columns)
```

```python
from sklearn.inspection import permutation_importance
```

```python
perm = permutation_importance(
    pipe_rf, X_val, y_val, n_repeats=10, random_state=42, scoring='r2'
)
```

```python
pd.Series(perm.importances_mean, index=X.columns)
```

```python
pd.Series(perm.importances_std, index=X.columns)
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
from scipy.stats import randint
params = {
    'reg__n_estimators': randint(100, 500),
    'reg__max_depth': [None, 10, 15, 20],
    'reg__min_samples_split': randint(2, 20)
}
```

```python
rs = RandomizedSearchCV(
    pipe_rf,
    params,
    n_iter=25,
    cv=KFold(n_splits=5, shuffle=True, random_state=42),
    scoring='neg_root_mean_squared_error',
    random_state=42,
    n_jobs=-1,
    refit=True
)
```

```python
rs.fit(X_train, y_train)
```

```python
rs.best_params_
```

```python
rs.best_score_
```

```python
best = rs.best_estimator_
y_test_pred = best.predict(X_test)
```

```python
mean_squared_error(y_test, y_test_pred)
```

```python
mean_absolute_error(y_test, y_test_pred)
```

```python
r2_score(y_test, y_test_pred)
```

```python
mean_absolute_percentage_error(y_test, y_test_pred)
```
