```python
# Simple example of the Decorator Pattern with descriptions,
# using a common abstract CondimentDecorator base class for milk and sugar
# Now with a Beverage base class, and an additional Latte class

from abc import ABC, abstractmethod

class Beverage(ABC):
    @abstractmethod
    def cost(self):
        pass

    @abstractmethod
    def description(self):
        pass

class Coffee(Beverage):
    def cost(self):
        return 5
    
    def description(self):
        return "Coffee"

class Latte(Beverage):
    def cost(self):
        return 7
    
    def description(self):
        return "Latte"

class CondimentDecorator(Beverage, ABC):
    def __init__(self, beverage):
        self.beverage = beverage

    @abstractmethod
    def cost(self):
        pass

    @abstractmethod
    def description(self):
        pass

class MilkDecorator(CondimentDecorator):
    def cost(self):
        return self.beverage.cost() + 2
    
    def description(self):
        return self.beverage.description() + " with milk"

class SugarDecorator(CondimentDecorator):
    def cost(self):
        return self.beverage.cost() + 1
    
    def description(self):
        return self.beverage.description() + " and sugar"

# Usage:
# Basic coffee
coffee = Coffee()
print("Cost of simple coffee:", coffee.cost())
print("Description:", coffee.description())

# Latte
latte = Latte()
print("Cost of simple latte:", latte.cost())
print("Description:", latte.description())

# Coffee with milk
milk_coffee = MilkDecorator(coffee)
print("Cost of coffee with milk:", milk_coffee.cost())
print("Description:", milk_coffee.description())

# Coffee with milk and sugar
milk_sugar_coffee = SugarDecorator(milk_coffee)
print("Cost of coffee with milk and sugar:", milk_sugar_coffee.cost())
print("Description:", milk_sugar_coffee.description())

# Latte with milk and sugar
milk_sugar_latte = SugarDecorator(MilkDecorator(latte))
print("Cost of latte with milk and sugar:", milk_sugar_latte.cost())
print("Description:", milk_sugar_latte.description())
```
