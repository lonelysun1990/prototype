```python
# Adapter Pattern Example

# Existing class that needs adapting
class OldPrinter:
    def print_text(self, text):
        print(f"OldPrinter: {text}")

# Target interface expected by the client
class NewPrinterInterface:
    def print_message(self, message):
        pass

# Adapter that makes OldPrinter compatible with NewPrinterInterface
class PrinterAdapter(NewPrinterInterface):
    def __init__(self, old_printer):
        self.old_printer = old_printer

    def print_message(self, message):
        self.old_printer.print_text(message)

# Client code
def client_code(printer: NewPrinterInterface):
    printer.print_message("Hello, Adapter Pattern!")

# Usage
old = OldPrinter()
adapter = PrinterAdapter(old)
client_code(adapter)
```

```python
# Facade Pattern Example

# Subsystem classes
class CPU:
    def freeze(self):
        print("Freezing processor.")

    def jump(self, position):
        print(f"Jumping to position {position}.")

    def execute(self):
        print("Executing instructions.")

class Memory:
    def load(self, position, data):
        print(f"Loading data '{data}' into position {position}.")

class HardDrive:
    def read(self, sector):
        print(f"Reading sector {sector}.")
        return f"Data from sector {sector}"

# Facade
class ComputerFacade:
    def __init__(self):
        self.cpu = CPU()
        self.memory = Memory()
        self.hard_drive = HardDrive()

    def start(self):
        print("Starting computer via Facade:")
        self.cpu.freeze()
        data = self.hard_drive.read(100)
        self.memory.load(100, data)
        self.cpu.jump(100)
        self.cpu.execute()

# Client code
if __name__ == "__main__":
    computer = ComputerFacade()
    computer.start()
```

```python
# Proxy Pattern Example

class RealSubject:
    def request(self):
        print("RealSubject: Handling request.")

class Proxy:
    def __init__(self, real_subject):
        self._real_subject = real_subject

    def request(self):
        print("Proxy: Doing some access control before forwarding to RealSubject.")
        self._real_subject.request()
        print("Proxy: Doing some post-processing after forwarding to RealSubject.")

# Client code
if __name__ == "__main__":
    real = RealSubject()
    proxy = Proxy(real)
    proxy.request()
```

```python
# Bridge Pattern Example

# Abstraction
class RemoteControl:
    def __init__(self, device):
        self._device = device

    def turn_on(self):
        self._device.power_on()

    def turn_off(self):
        self._device.power_off()

# Implementor
class Device:
    def power_on(self):
        pass

    def power_off(self):
        pass

# Concrete Implementors
class TV(Device):
    def power_on(self):
        print("Turning on the TV.")

    def power_off(self):
        print("Turning off the TV.")

class Radio(Device):
    def power_on(self):
        print("Turning on the radio.")

    def power_off(self):
        print("Turning off the radio.")

# Client code
if __name__ == "__main__":
    tv = TV()
    radio = Radio()
    remote_tv = RemoteControl(tv)
    remote_radio = RemoteControl(radio)

    remote_tv.turn_on()
    remote_tv.turn_off()

    remote_radio.turn_on()
    remote_radio.turn_off()
```
