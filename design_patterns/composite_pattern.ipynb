```python
# The composite pattern allows treating individual objects and compositions of objects uniformly.

from abc import ABC, abstractmethod

# Component
class Graphic(ABC):
    @abstractmethod
    def draw(self):
        pass

# Leaf
class Circle(Graphic):
    def draw(self):
        print("Drawing a Circle")

class Square(Graphic):
    def draw(self):
        print("Drawing a Square")

# Composite
class Picture(Graphic):
    def __init__(self):
        self.children = []

    def add(self, graphic):
        self.children.append(graphic)

    def remove(self, graphic):
        self.children.remove(graphic)

    def draw(self):
        print("Drawing a Picture containing:")
        for child in self.children:
            child.draw()

# Usage example
if __name__ == "__main__":
    circle1 = Circle()
    square1 = Square()

    # Create a composite
    picture = Picture()
    picture.add(circle1)
    picture.add(square1)

    # Add another composite inside (nested structure)
    nested_picture = Picture()
    nested_picture.add(Circle())
    picture.add(nested_picture)

    picture.draw()
```

```python

# Simple example of the Composite Pattern for ML model building in PyTorch

import torch
import torch.nn as nn

# Component
class Layer(nn.Module):
    def forward(self, x):
        raise NotImplementedError

# Leaf
class LinearLayer(Layer):
    def __init__(self, in_features, out_features):
        super().__init__()
        self.linear = nn.Linear(in_features, out_features)
    
    def forward(self, x):
        return self.linear(x)

class ReLULayer(Layer):
    def __init__(self):
        super().__init__()
        self.relu = nn.ReLU()
    
    def forward(self, x):
        return self.relu(x)

# Composite
class SequentialComposite(Layer):
    def __init__(self):
        super().__init__()
        self.children = nn.ModuleList()
    
    def add(self, layer):
        self.children.append(layer)
    
    def forward(self, x):
        for layer in self.children:
            x = layer(x)
        return x

# Usage Example
if __name__ == "__main__":
    # Build a simple model: Linear -> ReLU -> Linear
    model = SequentialComposite()
    model.add(LinearLayer(4, 8))
    model.add(ReLULayer())
    model.add(LinearLayer(8, 2))

    # Create a nested composite
    nested = SequentialComposite()
    nested.add(LinearLayer(2, 2))
    nested.add(ReLULayer())

    # Add nested composite to main model
    model.add(nested)

    # Test with random input
    x = torch.randn(1, 4)
    output = model(x)
    print("Output:", output)
```
