```python
# TF-IDF
```

```python
docs = [
    "cat cat dog",
    "cat mouse",
    "dog dog mouse"
]
```

```python
import numpy as np
```

```python
tokenized_docs = [doc.lower().split() for doc in docs]
```

```python
tokenized_docs
```

```python
n = len(tokenized_docs)
```

```python
from collections import defaultdict
df = defaultdict(int)
```

```python
for tokens in tokenized_docs:
    unique_words = set(tokens)
    for word in unique_words:
        df[word] += 1
```

```python
df
```

```python
res = []
```

```python
for tokens in tokenized_docs:
    word_counts = defaultdict(int)
    for token in tokens:
        word_counts[token] += 1

    total_words = len(tokens)

    doc_scores = {}
    for token, count in word_counts.items():
        tf = count / total_words
        tf_idf = tf * np.log(n / df[token])
        doc_scores[token] = tf_idf
    res.append(doc_scores)
```

```python
res
```

```python
# Bag of words
# get a matrix where each row is the word-count vector for one document
```

```python
docs = [
    "cat sat cat",
    "dog sat"
]
```

```python
tokenized_docs = [doc.lower().split() for doc in docs]
```

```python
tokenized_docs
```

```python
vocab = sorted(set([token for tokens in tokenized_docs 
    for token in tokens ]))
```

```python
vocab
```

```python
word_to_index = {word:i for i, word in enumerate(vocab)}
```

```python
word_to_index
```

```python
res = []
n = len(vocab)
for tokens in tokenized_docs:
    word_count = defaultdict(int)
    for token in tokens:
        word_count[word_to_index[token]] += 1
    vec = [0] * n
    for idx, count in word_count.items():
        vec[idx] = count
    res.append(vec)
```

```python
res
```
