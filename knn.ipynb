```python
# putting all together
def knn(query, data, y, k):
    n = data.shape[0]
    dists = np.zeros(n)
    for i in range(n):
        dists[i] = distance(data[i], query)
    # pick k neighbors
    neighbors = np.argsort(dists)[:k]

    # majority vote for classification
    from collections import defaultdict
    votes = defaultdict(int)
    target = 0
    target_count = 0
    for n in neighbors:
        votes[y[n]] += 1
        if votes[y[n]] > target_count:
            target = y[n]
            target_count = votes[y[n]]
    return target

def distance(p1, p2):
    # both are 1-dim np arrays
    return np.sqrt(np.sum((p1 - p2)**2))
```

```python
knn(query, X, y, k=3)
```

```python
# knn for regression
X = np.array([
    [1], [2], [3], [6], [7], [8]
])

y = np.array([2, 3, 2.5, 7, 8, 7.5])

query = np.array([4])
```

```python
def knn_r(X, y, query, k=3):
    n = X.shape[0]
    dists = np.zeros(n)
    for i in range(n):
        dists[i] = distance(X[i], query)
    # pick k neighbors
    neighbors = np.argsort(dists)[:k]
    
    # select a subset
    target = y[neighbors].mean()
    return target
    
def distance(p1, p2):
    return np.sqrt(np.sum((p1-p2)**2))
```

```python
target = knn_r(X, y, query)
```
