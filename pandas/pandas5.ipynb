```python
def valid_ip(x):
    octets = x.split('.')
    if len(octets) != 4:
        return False
    for o in octets:
        if o.startswith('0'):
            return False
        if int(o) > 255:
            return False
    return True
        
logs['valid'] = logs['ip'].apply(valid_ip)
```

```python
logs[logs.valid==False].groupby('ip')['log_id'].count()\
    .reset_index().rename(columns={'log_id': 'invalid_count'})\
    .sort_values(['invalid_count', 'ip'], ascending=False)
```

```python
from collections import defaultdict
levels = defaultdict(int)
budgets = defaultdict(int)
team_size = defaultdict(int)
```

```python
def dfs(emp_id, df, level):
    # print('emp_id: ', emp_id)
    if emp_id not in df.employee_id:
        return 0, 0 # team
    levels[emp_id] = level
    subs = df[df.employee_id == emp_id]
    if len(subs) == 1 and np.isnan(subs['sub_id'].values[0]):
        budgets[emp_id] = subs['salary'].values[0]
        team_size[emp_id] = 0
        return budgets[emp_id], team_size[emp_id] + 1

    for i in range(len(subs)):
        row = subs.iloc[i]
        sub_budget, sub_size = dfs(int(row['sub_id']), df, level+1)
        team_size[emp_id] += sub_size
        budgets[emp_id] += sub_budget
    return budgets[emp_id], team_size[emp_id] + 1
        
```
