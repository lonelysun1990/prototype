```python
# logistic regression
import numpy as np

X = np.array([
    [1.0, 2.0],
    [2.0, 1.0],
    [2.0, 3.0],
    [6.0, 5.0],
    [7.0, 8.0],
    [8.0, 6.0]
])

y = np.array([0, 0, 0, 1, 1, 1])
```

```python
n, d = X.shape
W = np.random.randn(d)
b = np.random.randn(1)
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
y_prob = sigmoid(X @ W + b)
```

```python
(y_prob > 0.5).astype(int)
```

```python
class LogisticRegression:
    def __init__(self, lr = 0.01, iter = 1000):
        self.W = None
        self.b = None
        self.lr = lr
        self.iter = iter

    def _sigmoid(self, z):
        return 1 / (1 + np.exp(-z))
    
    def fit(self, X, y):
        n, d = X.shape
        # initialize
        self.W = np.random.randn(d)
        self.b = np.random.randn(1)

        # update
        for _ in range(self.iter):
            y_prob = self._sigmoid(X @ self.W + self.b)
            dW = - 1 / n * X.T @ (y - y_prob)
            db = - 1 / n * np.sum(y - y_prob)

            self.W -= self.lr * dW
            self.b -= self.lr * db
            

    def predict(self, X):
        y_prob = self._sigmoid(X @ self.W + self.b)
        return (y_prob > 0.5).astype(int)
        
    def predict_proba(self, X):
        return self._sigmoid(X @ self.W + self.b)
```

```python
logistic_regression = LogisticRegression()
```

```python
logistic_regression.fit(X, y)
```

```python
y_pred = logistic_regression.predict(X)
```

```python
y_proba = logistic_regression.predict_proba(X)
```
